```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.cluster import KMeans
```

```python
data = {
    "Order_ID": [101, 102, 103, 104, 105, 106],
    "Distance_km": [12, 25, 8, 40, 30, 15],
    "Delivery_Time_hours": [1.5, 3.0, 1.0, 5.0, 4.0, 2.0],
    "Order_Value": [1200, 2500, 800, 4000, 3200, 1500],
    "Inventory_Level": [50, 30, 70, 20, 25, 60]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.describe())
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Distance_km"], df["Delivery_Time_hours"])
plt.xlabel("Distance (km)")
plt.ylabel("Delivery Time (hours)")
plt.title("Distance vs Delivery Time")
plt.show()
```

```python
X = df[["Distance_km"]]
y = df["Delivery_Time_hours"]

model = LinearRegression()
model.fit(X, y)

predictions = model.predict(X)

print("Predicted Delivery Times:")
print(predictions)
```

```python
features = df[["Distance_km", "Order_Value"]]

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(features)

print(df)
```
